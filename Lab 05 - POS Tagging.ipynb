**POS Tagging**

**NLTK Tagger**

A part-of-speech tagger, or POS-tagger, processes a sequence of words, and attaches a part of speech tag to each word.

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

text1 = word_tokenize("And now for something completely different")
print(nltk.pos_tag(text1))
print()

text2 = word_tokenize("They refuse to permit us to obtain the refuse permit")
print(nltk.pos_tag(text2))
print()

# to get the meaning of the tags
nltk.help.upenn_tagset('WDT')

text3 = word_tokenize("The back door.")
print(nltk.pos_tag(text3))
print()

text4 = word_tokenize("I couldn’t get back to sleep")
print(nltk.pos_tag(text4))
print()
```

**Regular Expression Tagger**

The regular expression tagger assigns tags to tokens on the basis of matching patterns. For instance, we might guess that any word ending in ed is the past participle of a verb, and any word ending with 's is a possessive noun. We can express these as a list of regular expressions:

```python
patterns = [
     (r'.*ing$', 'VBG'),               # gerunds
     (r'.*ed$', 'VBD'),                # simple past
     (r'.*es$', 'VBZ'),                # 3rd singular present
     (r'.*ould$', 'MD'),               # modals
     (r'.*\'s$', 'NN$'),               # possessive nouns
     (r'.*s$', 'NNS'),                 # plural nouns
     (r'^-?[0-9]+(.[0-9]+)?$', 'CD'),  # cardinal numbers
     (r'.*', 'NN'),                    # nouns (default)
     (r'^\d+$', 'CD'),
     (r'.*ing$', 'VBG'),               # gerunds, i.e. wondering
     (r'.*ment$', 'NN'),               # i.e. wonderment
     (r'.*ful$', 'JJ')                 # i.e. wonderful
 ]

regexp_tagger = nltk.RegexpTagger(patterns)
tagger = nltk.tag.sequential.RegexpTagger(patterns)

import nltk
from nltk.tokenize import word_tokenize
text1 = word_tokenize('Python is a high-level, general-purpose programming language')
print("Regular Expression Tagging ~~~ \n", tagger.tag(text1))
print()
print("NLTK POS Tagging ~~~ \n", nltk.pos_tag(text1)) # using NLTK
```

**spaCy Tagger**

Part-of-speech (POS) tagging with SpaCy is a common task in Natural Language Processing (NLP) that involves identifying the grammatical category of each word in a sentence, such as noun, verb, adjective, etc. SpaCy is a powerful NLP library that makes POS tagging easy to implement.

Common POS Tags in SpaCy:
~~~~~~~~~~~~~~~~~~~~~~~~~
PROPN = Proper noun
NOUN = Noun
VERB = Verb
ADJ = Adjective
ADV = Adverb
AUX = Auxiliary verb
DET = Determiner
ADP = Adposition (prepositions or postpositions)
PUNCT = Punctuation
PRON = Pronoun

```python
import spacy

# Load the SpaCy model
nlp = spacy.load("en_core_web_sm")

# Sample text for POS tagging
text = "Python is a high-level, general-purpose programming language. Python is a high-level, general-purpose programming language."

# Process the text through the model
doc = nlp(text)

# Iterate over the tokens in the processed text and print the word and its POS tag
for token in doc:
    print(token.text, "|", token.lemma_, "|", token.pos_, "|", token.tag_, "|", token.dep_, "|", token.is_alpha, "|", token.is_stop)
```

**POS Tagging using Transformers (Hugging Face)**

Hugging Face's Transformers library provides access to cutting-edge deep learning models like BERT, GPT, RoBERTa, etc., which can be used for POS tagging, although it’s typically overkill for simpler tasks.

```python
from transformers import pipeline
import warnings
warnings.filterwarnings("ignore")

# Load the POS tagging pipeline from Hugging Face
pos_tagger = pipeline("ner", model="vblagoje/bert-english-uncased-finetuned-pos")
text = "Python is a high-level, general-purpose programming language. Python is a high-level, general-purpose programming language."

# Run POS tagging
pos_tags = pos_tagger(text)

# Display the results
print(pos_tags)
```

**Lemmatization with POS Tags Specifications**

```python
import nltk
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer
def get_wordnet_pos(word):
    """Map POS tag to first character lemmatize() accepts"""
    tag = nltk.pos_tag([word])[0][1][0].upper()
    tag_dict = {"J": wordnet.ADJ,
                "N": wordnet.NOUN,
                "V": wordnet.VERB,
                "R": wordnet.ADV}

    return tag_dict.get(tag, wordnet.NOUN)


# Init Lemmatizer
lemmatizer = WordNetLemmatizer()

# Lemmatize Single Word with the appropriate POS tag
word = 'feet'
print(lemmatizer.lemmatize(word, get_wordnet_pos(word)))

# Lemmatize a Sentence with the appropriate POS tag
sentence = "The striped bats are hanging on their feet for the best"
print([lemmatizer.lemmatize(w, get_wordnet_pos(w)) for w in nltk.word_tokenize(sentence)])
```
