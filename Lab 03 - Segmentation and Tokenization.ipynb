**Word Segementation (Tokenization)**

**NLTK**

```python
from nltk.tokenize import word_tokenize

sampleText = "This programme 😎✨ is designed to provide students with knowledge and applied skills in data science, big data analytics and business intelligence. It aims to develop analytical and investigative knowledge and skills using data science tools and techniques, and to enhance data science knowledge and critical interpretation skills. Students will understand the impact of data science upon modern processes and businesses, be able to identify, and implement specific tools, practices, features and techniques to enhance the analysis of data."

Tokens = word_tokenize(sampleText)

print("There are", len(Tokens), "tokens in this text\n")

counter = 0
for w in Tokens:
    counter+=1
    print(counter,".",w)
    
# To print all the tokens
print("\n",Tokens)  
print()

# Use nltk.Text() to create a text list from the tokens list.
import nltk    
Tokenstext = nltk.Text(Tokens)
print(Tokenstext[0:len(Tokenstext)])
```

**Removing Stop Words & punctuation**

Removing stop words after tokenization is a super common preprocessing step in NLP, and there are solid reasons behind it.

1. Reduce noise in the text: Stop words like “the”, “is”, “and”, “of” occur very frequently but usually add little semantic meaning.
2. Improve model efficiency: Reduces vocabulary size, Lowers memory usage and Speeds up training and inference. This matters a lot for: Bag-of-Words, TF-IDF and Classical ML models (LogReg, SVM, Naive Bayes).
3. Better feature quality (especially for BoW / TF-IDF).
4. Reduce dimensionality: By removing stop words help to get Fewer features, Less sparsity and Better generalization (less overfitting). This is especially useful when working with small or medium-sized datasets.

When you should NOT remove stop words:

1. When working on Sentiment Analysis. Words like: not, no, never are often in stop-word lists but flip sentiment.
2. Transformer-based models (BERT, RoBERTa, etc.): Modern models Learn context, Handle stop words naturally and Often perform worse if you remove them.
3. Grammar- or syntax-aware tasks such as: POS tagging, Dependency parsing and Question answering, Stop words are crucial for structure.

```python
import nltk, string

text = "This programme is designed to provide students with knowledge and applied skills in data science, big data analytics and business intelligence. It aims to develop analytical and investigative knowledge and skills using data science tools and techniques, and to enhance data science knowledge and critical interpretation skills. Students will understand the impact of data science upon modern processes and businesses, be able to identify, and implement specific tools, practices, features and techniques to enhance the analysis of data."

text_lower = text.lower()
wordtokens = nltk.tokenize.word_tokenize(text_lower)
print("Tokens ~~~ \n", wordtokens)

print("\nTotal number of words in this text corpus is",len(wordtokens))

stopTokens = nltk.corpus.stopwords.words("english") + list(string.punctuation)
filteredTokens = []

for w in wordtokens:
    if w not in stopTokens :
        filteredTokens.append(w)

print("There are", len(filteredTokens), "words in this text after removing stop words\n")
print("Filtered Tokens ~~~ \n", filteredTokens)

print("\nStop Tokens ~~~ \n", stopTokens)

print("\nWord count or Bag of words ~~~ \n", wordtokens.count("skills"))

fdist1 = nltk.FreqDist(filteredTokens)
print("\nFrequency Distribution ~~~\n",fdist1.most_common(10))
```

**Word Cloud**

Word Cloud is a data visualization technique used for representing text data in which the size of each word indicates its frequency or importance. Significant textual data points can be highlighted using a word cloud. Word clouds are widely used for analyzing data from social network websites.

Benefits of a Word Cloud:
1. Instant visual insight
2. Excellent for exploratory data analysis (EDA)
3. Communicates results to non-technical stakeholders
4. Quick validation of preprocessing steps
5. Highlight domain-specific vocabulary

A word cloud is a powerful exploratory and communication tool—but a weak analytical one.

```python
import wordcloud
# %pip install wordcloud

import matplotlib.pyplot as plt
# Display the generated word cloud ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~
word_cloud = wordcloud.WordCloud(background_color='white').generate(text) #untokenized text
plt.figure(figsize = (12, 8), facecolor = None)
plt.imshow(word_cloud)
plt.axis("off")
plt.tight_layout(pad = 0)
plt.show()
```
